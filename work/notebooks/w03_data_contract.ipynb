```python
from google.colab import userdata
import duckdb

# Retrieve Hugging Face token from Colab Secrets
HF_TOKEN = userdata.get('HF_TOKEN')

con = duckdb.connect()

# Configure DuckDB HTTP filesystem to use the Hugging Face token
con.execute(f"""
CREATE OR REPLACE SECRET hf_auth (
    TYPE HTTP,
    BEARER_TOKEN '{HF_TOKEN}'
)
"""
)

query = """
SELECT
    COUNT(*) AS row_count,
    MIN(report_date) AS min_date,
    MAX(report_date) AS max_date
FROM read_parquet(
    'hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance/month=*/data_0.parquet'
)
"""

result = con.execute(query).fetchdf()
print("Row count and date range:")
print(result)

grain_check = """
SELECT
    report_date,
    client_hash_id,
    content_hash_id,
    COUNT(*) AS duplicate_count
FROM read_parquet(
    'hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance/month=*/data_0.parquet'
)
GROUP BY
    report_date,
    client_hash_id,
    content_hash_id
HAVING COUNT(*) > 1
LIMIT 5
"""

grain_result = con.execute(grain_check).fetchdf()
print("\nGrain duplicates check:")
print(grain_result)

duplicate_check = """
SELECT *
FROM read_parquet(
    'hf://datasets/FlyRank/internship-warehouse/fact_content_daily_performance/month=*/data_0.parquet'
)
WHERE report_date = '2026-06-13'
  AND client_hash_id = 'client_8ddc46da5414ffd8'
  AND content_hash_id = 'content_1d06a2c99a935e49'
"""

duplicate_rows = con.execute(duplicate_check).fetchdf()
print("\nDuplicate rows:")
print(duplicate_rows)
```
